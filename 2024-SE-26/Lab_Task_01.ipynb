```python
Exercise2: if,elif,and else
```

```python
intensities=[-1,0,127,128,255,256]

for intensity in intensities:
    if intensity<0 or intensity>255:
        print(intensity,"Invalid")
    elif intensity>=128:
        print(intensity,"Bright pixel")
    else:
        print(intensity,"Darkpixel")
```

```python
Exercise3: for and whileloop
```

```python
pixels=[20,130,255]

for pixel in pixels:
    if pixel >=128:
        print(pixel,"Bright pixel")
    else:
        print(pixel,"Dark pixel")

count=0

while count<3:
    print(count)
    count=count+1

        
```

```python
Exercise4:Function
```

```python
def is_bright(pixel,threshold=128):
    if pixel >=threshold:
        return True
    else:
        return False
print(is_bright(127))
print(is_bright(128))
print(is_bright(200,210))
```
